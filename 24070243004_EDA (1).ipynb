```python
# Univariate Analysis
# For Numerical Variables

import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(df['PREV.CLOSE'], bins=30, kde=True)  # Histogram
plt.show()

sns.boxplot(x=df['PREV.CLOSE'])  # Boxplot for outliers
plt.show()
```

```python
# Univariate Analysis - Closing Price Trend
plt.figure(figsize=(10, 5))
sns.lineplot(x=df["Date"], y=df["close"], marker="o", label="Closing Price", color="blue")
plt.xlabel("Date")
plt.ylabel("Closing Price (INR)")
plt.title("Stock Closing Price Trend")
plt.xticks(rotation=45)
plt.legend()
plt.grid()
plt.show()
```

```python
# Volume Trends
plt.figure(figsize=(10, 5))
sns.lineplot(x=df["Date"], y=df["VOLUME"], marker="o", label="Trading Volume", color="red")
plt.xlabel("Date")
plt.ylabel("Volume")
plt.title("Stock Trading Volume Over Time")
plt.xticks(rotation=45)
plt.legend()
plt.grid()
plt.show()
```

```python
# Bivariate Analysis - Scatter Plot
plt.figure(figsize=(6, 4))
sns.scatterplot(x=df["OPEN"], y=df["PREV. CLOSE"], color="green")
plt.xlabel("Opening Price")
plt.ylabel("Previous Close Price")
plt.title("Opening Price vs Previous Close Price")
plt.show()
```

```python
# Additional Analysis: Price Volatility
plt.figure(figsize=(10, 5))
df["Volatility"] = df["HIGH"] - df["LOW"]
sns.lineplot(x=df["Date"], y=df["Volatility"], marker="o", label="Price Volatility", color="purple")
plt.xlabel("Date")
plt.ylabel("Volatility (High - Low)")
plt.title("Stock Price Volatility Over Time")
plt.xticks(rotation=45)
plt.legend()
plt.grid()
plt.show()
```

```python
# Correlation Heatmap
numeric_df = df.select_dtypes(include=['number'])  # Select only numeric columns
plt.figure(figsize=(10, 6))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Matrix")
plt.show()
```

```python
# SUMMARIZE

# Step 1: Clean Column Names (Remove Spaces)
df.columns = df.columns.str.strip()

# Step 2: Convert Date to datetime format
df["Date"] = pd.to_datetime(df["Date"], format="%d-%b-%Y")

# Step 3: Convert VOLUME, VALUE, No of trades to numeric (removing commas)
df["VOLUME"] = df["VOLUME"].str.replace(",", "").astype(float)
df["VALUE"] = df["VALUE"].str.replace(",", "").astype(float)
df["No of trades"] = df["No of trades"].str.replace(",", "").astype(int)

# Step 4: Check for missing values
missing_values = df.isnull().sum()

# Step 5: Check for duplicates
duplicate_rows = df.duplicated().sum()

# Display cleaned data info
df.info(), missing_values, duplicate_rows
```

```python
# Fix for VOLUME column
if df['VOLUME'].dtype == 'object':
    df['VOLUME'] = df['VOLUME'].str.replace(',', '')
    df['VOLUME'] = pd.to_numeric(df['VOLUME'], errors='coerce')

# Fix for No of trades column
if df['No of trades'].dtype == 'object':
    df['No of trades'] = df['No of trades'].str.replace(',', '')
    df['No of trades'] = pd.to_numeric(df['No of trades'], errors='coerce')
```

PIVOT TABLE FUNCTION

```python
# 1. Average Closing Price by Weekday
df['Date'] = pd.to_datetime(df['Date'])
df['Weekday'] = df['Date'].dt.day_name()
pivot1 = df.pivot_table(values='close', index='Weekday', aggfunc='mean')
print(pivot1)

```

```python
# 2. Total Volume Traded by Date
# df['VOLUME'] = df['VOLUME'].str.replace(',', '').astype(int)
pivot2 = df.pivot_table(values='VOLUME', index='Date', aggfunc='sum')
print(pivot2)
```

```python
# 3. Max and Min Prices by Weekday

pivot3 = df.pivot_table(values=['HIGH', 'LOW'], index='Weekday', aggfunc={'HIGH': 'max', 'LOW': 'min'})
print(pivot3)
```

```python
# 4. Average VWAP by Weekday
pivot4 = df.pivot_table(values='vwap', index='Weekday', aggfunc='mean')
print(pivot4)
```

```python
# 5. Number of Trades per Weekday

# df['No of trades'] = df['No of trades'].str.replace(',', '').astype(int)
pivot5 = df.pivot_table(values='No of trades', index='Weekday', aggfunc='sum')
print(pivot5)
```
